#  Seaborn Library of Matplotlib
- Seaborn is a data visualization library in matplotlib that provides high level interface for creating interactive and more colorful. Plots with less code

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Retrieves the default qualitative color palette from Seaborn (usually 'tab10')
palette = sns.color_palette()
print(palette)
# Visually displays a strip of the colors in the palette
sns.palplot(palette)
```

```python
# Retrieves the sequential 'Greens' palette with 20 shades.
# Sequential palettes are best for data ranging from low to high.
palette_green = sns.color_palette('Greens', 20)
# Visually displays the strip of the 'Greens' color palette
sns.palplot(palette_green)
```

```python
data_dict = {
    'building':['burj-khalifa', 'medrika','eiffil-tower'],
    'category':['highest', 'medium','low'],
    'building_height': [150, 120, 100]
}
# Custom list of colors to be applied to the bars
bar_colors = ['red', 'green', 'yellow']
# Converts the dictionary into a Pandas DataFrame, the standard format for Seaborn
data = pd.DataFrame(data_dict)
# Creates a Bar Plot using Seaborn:
sns.barplot(
    x = 'building', # Categorical variable for the x-axis labels
    y = 'building_height', # Numerical variable for the y-axis (bar height)
    data = data, # Specifies the DataFrame source
    hue = 'category', # Colors and groups the bars by the 'category' variable
    palette = bar_colors # Applies the custom color list
)
plt.show()
```

```python
# Loads the classic 'iris' flower dataset directly into a Pandas DataFrame
data_iris = sns.load_dataset('iris')
data_iris
# data_flight = sns.load_dataset('flights')
# data_flight
```

```python
sns.barplot(
    x = 'species', # Species categories on the x-axis
    y = 'sepal_width', # Mean sepal width on the y-axis (includes confidence interval lines)
    data = data_iris, 
    hue = 'species', # Ensures distinct colors for each species bar
    palette = bar_colors
)
plt.title('Iris Species Distribution Bar Chart')
plt.xlabel('Name Species')
plt.ylabel('Sepal Width')
plt.grid(axis = 'y', linestyle = '--', alpha = 0.5, color = 'lightblue')
plt.show()
```

```python
# Creates a Violin Plot, which shows the full probability distribution (density) of 'sepal_width'
# The width of the 'violin' shape reflects the data density at that value.

sns.violinplot(x = 'species', y = 'sepal_width', data = data_iris, hue = 'species', palette = bar_colors)
plt.title('Iris Species Distribution Violin Chart')
plt.xlabel('Name Species')
plt.ylabel('Sepal Width')
plt.grid(axis = 'y', linestyle = '--', alpha = 0.5, color = 'lightblue')
plt.show()
```

```python
sns.violinplot(x = 'species', y = 'sepal_width', data = data_iris, hue = 'species', palette = 'Set2')
plt.title('Iris Species Distribution Violin Chart')
plt.xlabel('Name Species')
plt.ylabel('Sepal Width')
plt.grid(axis = 'y', linestyle = '--', alpha = 0.5, color = 'lightblue')
plt.show()
```

```python
sns.stripplot(x = 'species', y = 'sepal_width', data = data_iris, hue = 'species', palette = bar_colors)
plt.title('Iris Species Distribution Strip Chart')
plt.xlabel('Name Species')
plt.ylabel('Sepal Width')
plt.grid(axis = 'y', linestyle = '--', alpha = 0.5, color = 'lightblue')
plt.show()
```

```python
data_iris = sns.load_dataset('iris')
sns.pairplot(data = data_iris, hue = 'species')
plt.show()
```

```python
sns.set_style('darkgrid')
sns.scatterplot(x = 'sepal_length', y = 'sepal_width', hue = 'species', data = data_iris)
plt.title('Darkgrid theme Example using Iris Data')
plt.show()
```

```python
sns.set_style('ticks')
sns.scatterplot(x = 'sepal_length', y = 'sepal_width', hue = 'species', data = data_iris)
plt.title('Darkgrid theme Example using Iris Data')
plt.xlabel('Sepal length')
plt.ylabel('Sepal Width')
plt.xticks([4,5,6,7,8], fontsize = 10, color = 'darkblue')
plt.yticks([1,2,3,4,5,6,7], fontsize = 10, color = 'purple')
sns.despine()
plt.show()
```
