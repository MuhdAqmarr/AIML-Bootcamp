# Matplotlib.pyplot.subplots()
- subplots(): provides a way to plot multiple plots  on single figure. 
- Given number of rows and columns . It returns a tuple (fig,ax) giving a single figure fig with an array of axes ax.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
fig,((ax1,ax2),(ax3,ax4))=plt.subplots(nrows=2,ncols=2,figsize=(10,5))
```

```python
sales_data = {'Electronic':45000,'Clothing':30000,'Grocery':25000,'Mobile':20000}
category = list(sales_data.keys())
actual_sale = list(sales_data.values())
print(category)
print(actual_sale)
fig, ((ax1,ax2),(ax3,ax4))=plt.subplots(nrows = 2, ncols = 2, figsize = (10,5))
x = np.random.random(10)
ax1.plot(x,x/2)
ax2.scatter(np.random.random(10), np.random.random(10))
ax3.bar(category, actual_sale)
ax4.hist(np.random.randn(1000))
```

## Subplots using inndexing

```python
our_lines = np.linspace(0,10,100)
product_prices = {'Hair Oil':20,'Face Wash':30,'Lip Balm':10,'Face Cream':15}
our_numbers = np.random.random(10)
our_dots = np.random.randn(1000)

# Config Chart Data
fig, ax = plt.subplots(nrows = 2, ncols = 2, figsize = (10, 8)) # saiz figure
ax[0,0].plot(our_lines,our_lines/2, color='#1f77b4', linewidth=2.5) # Set warna dan linewidth untuk Line Plot
ax[0,1].bar(list(product_prices.keys()), list(product_prices.values()), 
            color=['#ff7f0e', '#2ca02c','#d62728', '#9467bd'], edgecolor='black', alpha=0.9) # Tambah edgecolor dan alpha
ax[1,0].scatter(np.random.randn(100), np.random.randn(100), color = '#e377c2', alpha=0.6, s=40) # Guna alpha dan size (s) untuk Scatter
ax[1,1].hist(our_dots, color = '#17becf', bins=30, edgecolor='black', alpha=0.8) # Guna bins dan edgecolor untuk Hist

# Main Title
fig.suptitle('Matplotlib Subplots Example Using 2 Rows and 2 Cols', fontsize=16, fontweight='bold', y=1.02) # Besarkan font dan gerakkan sedikit ke atas

# Line Plot
ax[0,0].set_title('Line Plot Example', fontsize=13)
ax[0,0].set_xlabel('X Axis')
ax[0,0].set_ylabel('Y Axis')
ax[0,0].grid(True, linestyle = '-', alpha = 0.5, color = 'grey', axis = 'y') # grid solid line '-'

# Product Sale Info
ax[0,1].set_title('Product Sale Info', fontsize=13)
ax[0,1].set_xlabel('X Axis')
ax[0,1].set_ylabel('Y Axis')
ax[0,1].grid(True, linestyle = '-', alpha = 0.5, color = 'grey', axis = 'y') # grid solid line '-'
ax[0,1].tick_params(axis='x', rotation=15) # Rotate X-labels supaya tidak bertindih

# Scatter
ax[1,0].set_title('Scatter Example', fontsize=13)
ax[1,0].set_xlabel('X Axis')
ax[1,0].set_ylabel('Y Axis')
ax[1,0].grid(True, linestyle = '-', alpha = 0.5, color = 'grey', axis = 'y') # grid solid line '-'

# Histogram
ax[1,1].set_title('Histogram Example', fontsize=13)
ax[1,1].set_xlabel('X Axis')
ax[1,1].set_ylabel('Y Axis')
ax[1,1].grid(True, linestyle = '-', alpha = 0.5, color = 'grey', axis = 'y') # grid solid line '-'

plt.tight_layout()
plt.savefig('Subplotex.jpeg',dpi = 300, bbox_inches = 'tight')
```
