```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv('netflix_data.csv')
df.head(3)
```

```python
sns.set_theme(style = 'darkgrid', palette = 'Set2')
sns.countplot(data = df, x = 'type')
plt.title('Number of Movies vs Number of TV Shows')
plt.xlabel('Comtent Type')
plt.ylabel('Count')
plt.show()
```

```python
df = df.dropna(subset = ['type','genre'])
df['type']
```

```python
df['genre']
```

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df.head(10),x='type',hue='country',palette='Set2')
plt.title('Netflix Content Type by Top 10 Country' )
plt.xlabel('Content Type' )
plt.ylabel('Count')
plt.legend(title = 'Country')
plt.show()
```

```python
plt.figure(figsize = (10,5))
sns.countplot(data=df,x='type',hue='country',palette='Set2')
plt.title('Netflix Content Type by Country' )
plt.xlabel('Content Type' )
plt.ylabel('Count')
plt.legend(title = 'Country', ncol = 5, bbox_to_anchor = (1.05, 1), loc = 'upper left', frameon = True)
plt.show()
```

```python
data_iris = sns.load_dataset('iris')
plt.figure(figsize = (10,5))
sns.rugplot(data = data_iris)
plt.show()
```

```python
print(data_iris)
sns.kdeplot(data = data_iris, x = 'sepal_length', y = 'sepal_width')
plt.show()
```

## Use Netflix dataset to visualize the distribution of release years for movies vs tv show using seaborn's histplot

```python
plt.figure(figsize=(10, 5))
sns.histplot(
    data=df,
    x='release_year',
    hue='type',
    palette='Set1'
)

plt.title('Distribution of Content Release Years (Movies vs. TV Shows)')
plt.xlabel('Release Year')
plt.ylabel('Number of Release')
# plt.legend(title = 'Country', ncol = 5, bbox_to_anchor = (1.05, 1), loc = 'upper left')
plt.show()
```
