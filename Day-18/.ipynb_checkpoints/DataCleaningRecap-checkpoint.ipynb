```python
# Step 1 : downlaod and load dataset from the given link in chat & print dataset
df_stud = pd.read_csv('students_performance_dirty.csv')
df_stud
```

```python
# Step 2 : Get basic information about dataset
df_stud.info()
```

```python
# Step 3 : print('Missing values per column:')
print('Missing values per column:')
df_stud.isnull().sum()
```

```python
# Step 4 : Check missing data line by line
for i, row in df_stud.iterrows():
    print(f'Row {i}: Missing= {row.isnull().any()}, Details= {row.isnull().to_dict()}')
    
# missing_rows_stud = df_stud[df_stud.isnull().any(axis=1)]
# print('Rows with missing data: ')
# missing_rows_stud
```

```python
# Step 5 : Drop missing rows (if any)
df_cleaned_stud = df_stud.dropna()
print('After dropna(): ')
print('Shape before dropna(): ', df_stud.shape)
print('Shape after dropna(): ', df_cleaned_stud.shape)

df_cleaned_stud
```

```python
# Step 6 : Compare before and after
# Before Cleaning
print('Before dropna()')
df_stud
```

```python
# Step 6 : Compare before and after
print('after dropna()')
df_cleaned_stud
```

```python
# Step 7 : Fill Missing Value
# Fill missing value
df_filled_stud = df_stud.fillna({
    'gender' : 'Unknown',
    'study_hours' : round(df_stud['study_hours'].mean(), 1),
})
```

```python
# Step 8 : Compare after fillba()
# After filled missing value
for i in range(len(df_stud)):
    print(f'\nRow {i} Before:  {df_stud.iloc[i].to_dict()}')
    print(f'Row {i} Before:  {df_filled_stud.iloc[i].to_dict()}')
```

```python
# Identify rows where any of the critical columns are outside the [0, 100] range
# Note: attendance_pct has 91.2 which is fine, but 150.0 is an outlier.
impossible_mask = (df_filled_stud['attendance_pct'] > 100) | \
                  (df_filled_stud['attendance_pct'] < 0) | \
                  (df_filled_stud['math_score'] > 100) | \
                  (df_filled_stud['math_score'] < 0) | \
                  (df_filled_stud['reading_score'] > 100) | \
                  (df_filled_stud['reading_score'] < 0)

# Display the rows with impossible data
print("Impossible Data Rows:")
print(df_filled_stud[impossible_mask])
```

```python
import numpy as np

# Columns to be cleaned
score_cols = ['attendance_pct', 'math_score', 'reading_score']

# Define the threshold for impossible values
lower_bound = 0
upper_bound = 100

# 2a. Replace impossible values with NaN
for col in score_cols:
    # Identify values outside [0, 100] and replace them with NaN
    df_filled_stud.loc[(df_filled_stud[col] < lower_bound) | (df_filled_stud[col] > upper_bound), col] = np.nan

# 2b. Calculate the MEDIAN of the remaining (valid) data
median_values = df_filled_stud[score_cols].median()

# 2c. Impute the new NaN values (which were the impossible values) with the column median
df_filled_stud = df_filled_stud.fillna(median_values)

# Final check of a cleaned row (e.g., index 4, which had -10.0 attendance)
print("\nCleaned Row Example (Index 4):")
print(df_filled_stud.loc[4])
```
