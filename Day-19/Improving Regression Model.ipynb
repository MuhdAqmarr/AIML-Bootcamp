```python
import numpy as np
import pandas as pd
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
data = {
    'StudyHours': [2, 3, 4, 5, 6, 7, 8, 9, 10],
    'Attendance': [60, 65, 70, 72, 75, 80, 85, 88, 90],
    'PrevScore': [40, 45, 50, 55, 60, 65, 70, 75, 80],
    'FinalMarks': [50, 55, 60, 65, 70, 75, 78, 85, 88]
}

df=pd.DataFrame(data)
df
```

```python
y=df['FinalMarks']
y
```

```python
y_pred_baseline=np.repeat(y.mean(), len(y))
```

```python
mae=mean_absolute_error(y,y_pred_baseline)
print(mae)
```

```python
mse=mean_squared_error(y,y_pred_baseline)
print(mse)
```

```python
rmse=np.sqrt(mse)
print(rmse)
```

```python
r2= r2_score(y,y_pred_baseline)
print(f'Baseline MAE={mae:.2f}, RMSE{rmse:.2f}, R-Squared={r2:.2f}')
```

```python
# Clean the data
print(df.isnull().sum())
```

```python
df=df.fillna(df.mean())
```

```python
df
```

```python
# Data prep, find imput, output
X=df[['StudyHours','Attendance','PrevScore']]
y=df['FinalMarks']
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
# model creation
model=LinearRegression()
```

```python
# train the model
model.fit(X_train,y_train)
```

```python
print('Original Final Marks')
print(y)
y_all_prediction=model.predict(X)
print('Predicted all final Marks')
print(y_all_prediction)
```

```python
y_pred=model.predict(X_test)
print(y_pred)
```

```python
mae=mean_absolute_error(y_test,y_pred)
print(mae)
```

```python
mse=mean_squared_error(y_test,y_pred)
print(mse)
```

```python
rmse=np.sqrt(mse)
print(rmse)
```

```python
r2=r2_score(y_test,y_pred)
print(r2)
```

```python
print(f'Baseline MAE={mae:.2f}, RMSE{rmse:.2f}, R-Squared={r2:.2f}')
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
from sklearn.tree import DecisionTreeRegressor
```

```python
data = {
    'Area': [850, 900, 1000, 1100, 1200, 1500, 1600, 1800, 2000],
    'Bedrooms': [1, 2, 2, 2, 3, 3, 2, 4, 4],
    'Age': [1, 1, 3, 2, 1, 2, 2, 1, 2],
    'PriceIn100K': [5, 6, 7, 7, 7, 9, 9, 11, 7.]}
```

```python
df = pd.DataFrame(data)
```

```python
X = df[['Area', 'Bedrooms', 'Age']]
y = df['PriceIn100K']
```

```python
dt_model = DecisionTreeRegressor(min_samples_leaf=2, random_state=42)
```

```python
dt_model.fit(X, y)
```

```python
y_pred = dt_model.predict(X)
```

```python
mae = mean_absolute_error(y, y_pred)
mse = mean_squared_error(y, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y, y_pred)

print("--- Model Performance (min_samples_leaf=2) ---")
print(f'R-Squared (R2) = {r2:.2f}')
print(f'MAE = {mae:.2f}')
print(f'RMSE = {rmse:.2f}')
```
