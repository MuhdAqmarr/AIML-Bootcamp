# We are going to write one Multi Label Classification example using TensorFlow here.

```python
import tensorflow as tf
import pandas as pd
from sklearn.preprocessing import MultiLabelBinarizer
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
data = {
    "Overview":
    [
        # Original Samples
        # "A Hero Saves the city from destruction",
        # "A girl and boy become friends during war",
        # "A funny story of collegues during office meet",
        # "One scary nights in a deserted place",

        # Improvised Samples (Multi-Label focus)
        "A detective chases a supernatural killer through a gritty futuristic metropolis",
        "A high school prodigy accidentally invents a time machine and chaos ensues",
        "Two estranged sisters must fight a shadowy government conspiracy using ancient martial arts",
        "An eccentric chef enters a high-stakes cooking competition with hilarious and disastrous results",
        "A heartbreaking tale of a family separated during a global pandemic, battling isolation and hope",
        "A group of scientists uncovers a monstrous organism deep in the Antarctic ice",
        "A washed-up boxer trains for one last championship fight while juggling fatherhood and debt",
        "A medieval knight wakes up in the modern world and tries to navigate traffic and technology",
    ],
    "genres":
    [
        # Original Labels
        # ['Action', 'Adventure'],
        # ['Romance', 'Drama'],
        # ['Comedy'],
        # ['Horror'],

        # Improvised Labels
        ['Sci-Fi', 'Thriller', 'Crime'],
        ['Sci-Fi', 'Comedy', 'Adventure'],
        ['Action', 'Thriller', 'Drama'],
        ['Comedy', 'Family'],
        ['Drama', 'Romance'],
        ['Horror', 'Sci-Fi', 'Mystery'],
        ['Sport', 'Drama'],
        ['Fantasy', 'Comedy', 'Adventure'],
    ]
}
```

```python
df = pd.DataFrame(data)
df
```

```python
mlb = MultiLabelBinarizer()
y = mlb.fit_transform(df['genres'])
y
```

```python
tokenizer = Tokenizer(num_words=5000, oov_token='<OOV>')
tokenizer.fit_on_texts(df['Overview'])
X = tokenizer.texts_to_sequences(df['Overview'])
X = pad_sequences(X, maxlen=20, padding='post', truncating='post')
```

```python
model = tf.keras.Sequential(
    [
        tf.keras.layers.Embedding(input_dim=5000, output_dim=16, input_length=20),
        tf.keras.layers.GlobalAveragePooling1D(),
        tf.keras.layers.Dense(16, activation='relu'),
        tf.keras.layers.Dense(len(mlb.classes_), activation='sigmoid')
    ]
)
```

```python
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
model.fit(X, y, epochs=20)
```

```python
#evaluate
loss, accuracy = model.evaluate(X, y)
print(f"Loss: {loss:.4f}")
print(f"Accuracy: {accuracy:.4f}")
```

```python
# predict on a new movie overview
new_movie_overview = ['A detective chases a supernatural killer through a gritty futuristic metropolis']
seq = tokenizer.texts_to_sequences(new_movie_overview)
seq = pad_sequences(seq, maxlen=20, padding='post', truncating='post')
prediction = model.predict(seq)
prediction_labels = [mlb.classes_[i] for i in range(len(mlb.classes_)) if prediction[0][i] > 0.5]
print(f"Predicted genres: {prediction_labels}")
```

```python
data_2 = {
    "headline": [
        "Government launches new digital economy policy",
        "Star player wins championship after dramatic final",
        "Tech company releases AI-powered smartphone",
        "Stock market drops amid political tensions",
        "New environmental policy gains global support",
        "Space agency announces manned Mars mission"
    ],
    "topics": [
        ["Politics", "Economy", "Technology"],
        ["Sports"],
        ["Technology", "Business"],
        ["Economy", "Politics"],
        ["Environment", "Politics"],
        ["Science", "Technology"]
    ]
}
```

```python
df_2 = pd.DataFrame(data_2)
df_2
```

```python
mlb_2 = MultiLabelBinarizer()
y_2 = mlb_2.fit_transform(df_2['topics'])
y_2
```

```python
tokenizer_2 = Tokenizer(num_words=5000, oov_token='<OOV>')
tokenizer_2.fit_on_texts(df_2['headline'])
X_2 = tokenizer_2.texts_to_sequences(df_2['headline'])
X_2 = pad_sequences(X_2, maxlen=20, padding='post', truncating='post')
```

```python
model_2 = tf.keras.Sequential(
    [
        tf.keras.layers.Embedding(input_dim=5000, output_dim=16, input_length=20),
        tf.keras.layers.GlobalAveragePooling1D(),
        tf.keras.layers.Dense(16, activation='relu'),
        tf.keras.layers.Dense(len(mlb_2.classes_), activation='sigmoid')
    ]
)
```

```python
model_2.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
model_2.fit(X_2, y_2, epochs=20)
```

```python
#evaluate
loss, accuracy = model_2.evaluate(X_2, y_2)
print(f"Loss: {loss:.4f}")
print(f"Accuracy: {accuracy:.4f}")
```

```python
# predict on a new headline
new_headline = ['Star player wins championship after dramatic final']
seq2 = tokenizer_2.texts_to_sequences(new_headline)
seq2 = pad_sequences(seq2, maxlen=20, padding='post', truncating='post')
prediction2 = model_2.predict(seq2)
prediction_labels_2 = [mlb_2.classes_[i] for i in range(len(mlb_2.classes_)) if prediction2[0][i] > 0.5]
print(f"Predicted topics: {prediction_labels_2}")
```
