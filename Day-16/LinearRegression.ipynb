# Supervised learning

## Linear Regression

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
import numpy as np
```

```python
data = pd.read_csv('sales_data.csv')
data.head()
```

```python
x = data[['Advertising_Spend']]
y = data['Sales_Revenue']
```

```python
model = LinearRegression()
model.fit(x,y)
# print('Training done')
```

```python
new_spend = float(input('Enter New spend to predict sales revenue: '))
predicted_revenue = model.predict([[new_spend]])
print(f'Predicted sales revenue for RM{new_spend} ad spend: RM{predicted_revenue}')
```

## 1. Linear Regression - House price prediction 
- Goal: Train a Linear Regression model to predict house price from Size_sqft

```python
df_house = pd.read_csv('house_price.csv')
df_house.head(6)
```

```python
df_house['Price'] = df_house['Price'].replace('?', np.nan)
df_house['Price'] = pd.to_numeric(df_house['Price'], errors='coerce')
```

```python
df_house_cleaned = df_house.dropna(subset=['Size_sqft', 'Price'])
```

```python
size_sqft = df_house_cleaned[['Size_sqft']]
price = df_house_cleaned['Price']
```

```python
house_model = LinearRegression()
house_model.fit(size_sqft,price)
# print('Training done')
```

```python
sqft_new = float(input('Enter sqft to predict house price: '))
predicted_price = house_model.predict([[sqft_new]])
print(f'Predicted house price for {sqft_new} is: RM{predicted_price}')
```
