# KNN Algorithm in Supervised ML
## KNeighborsClassifier of sklearn library for this one

```python
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
```

```python
x = np.array([
    [1 ,3500],
    [2, 3200],
    [3, 3000],
    [4, 2500],
    [5, 2300],
    [6, 2000],
    [7, 2200],
    [2, 3700]
])

y = ['Unfit','Unfit','Unfit','Fit','Fit','Fit','Fit','Unfit']
```

```python
#Create a model
model = KNeighborsClassifier(n_neighbors = 3)
```

```python
#Train the model
model.fit(x,y)
```

```python
#Predict
new_person = [[4.5, 2600]]
predict = model.predict(new_person)
print('Predicted Result: ', predict[0])
```

```python
#Prediction from taking the values from users

exercise_hr = float(input('Enter number of exercise in hours: '))
cal_intake = float(input('Enter number of calories consume: '))
pred_fu = model.predict([[exercise_hr, cal_intake]])
print(f'Exercise Hours: {exercise_hr}, Calories Intake: {cal_intake}, So you are likely {pred_fu[0]}')
```

## Predict a student's grade level (A,B,C)
- Based on hours studied and assigment completed
- usinh K-Nearest Neighbors Algorithm

```python
df_stud_grade = pd.read_csv('student_grade.csv')
df_stud_grade
```

```python
x_stud = df_stud_grade[['Hours_Studied','Assignments_Completed']]
y_stud = df_stud_grade['Grade']
```

```python
#Create Model
model_stud = KNeighborsClassifier(n_neighbors = 3)
model_stud.fit(x_stud,y_stud)
```

```python
#Predict
new_stud = [[1.5,1]]
pred_gred = model_stud.predict(pd.DataFrame(new_stud, columns=['Hours_Studied', 'Assignments_Completed']))
print('Grade', pred_gred[0])
```
