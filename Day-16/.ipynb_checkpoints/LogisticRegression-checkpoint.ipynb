# 2. Logistic Regression - Pass/Fail

```python
import numpy as np
from sklearn.linear_model import LogisticRegression
import pandas as pd
```

```python
data = pd.read_csv('pass_fail.csv')
data.head()
```

```python
x = data[['Hours']]
y = data['Result']
```

```python
model = LogisticRegression()
model.fit(x,y)
```

```python
hours_input = float(input('Enter hours to predict result(Pass/Fail): '))
predicted_result = model.predict([[hours_input]])
print(f'Predicted result for {hours_input} hour of studies is: {predicted_result[0]}')
```

```python
pred_35 = model.predict([[3.5]])
print(f'Predicted result for {hours_input} hour of studies is: {pred_35[0]}')
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv('customer_purchase.csv')
df
```

```python
# Data split and labelling
le = LabelEncoder()
df['Purchased'] = le.fit_transform(df['Purchased'])
xCustomer_purchased = df[['Age','Salary']]
yCustomer_purchased = df['Purchased']
```

```python
df
```

```python
model_customer_purchased = LogisticRegression()
model_customer_purchased.fit(xCustomer_purchased, yCustomer_purchased)
```

```python
# prediction: predict the result using input from user
new_cust_age = int(input('Enter Age: '))
new_cust_sal = int(input('Enter Salary: '))
new_cust = [[new_cust_age,new_cust_sal]]
prediction_customer_purchased = model_customer_purchased.predict(new_cust)
print(f'Predicted Result (1=Yes, 0=No): {prediction_customer_purchased[0]}')
```

```python
# prediction: predict the result using input from user and display it with condition (if,else)
new_cust_age = int(input('Enter Age: '))
new_cust_sal = int(input('Enter Salary: '))
new_cust = [[new_cust_age,new_cust_sal]]
prediction_customer_purchased = model_customer_purchased.predict(new_cust)
if (prediction_customer_purchased):
    print('Likely purchased')
else:
    print('Likely not purchased')
```

- Use Logistic Regression Algorithm to trained the model
- Predict result if the email is spam or not

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
```

```python
df_spam_email = pd.read_csv('spam_dataset.csv')
df_spam_email
```

```python
print('Number of Missing values are:',df_spam_email.isnull().sum())
```

```python
xSpam = df_spam_email['message_content']
ySpam = df_spam_email['is_spam']

# Split the data (e.g., 80% for training, 20% for testing)
X_train_text, X_test_text, y_train, y_test = train_test_split(
    xSpam, 
    ySpam, 
    test_size=0.2, 
    random_state=42
)
print("Training samples:", len(X_train_text))
print("Testing samples:", len(X_test_text))

vectorizer = CountVectorizer()
X_train_numeric = vectorizer.fit_transform(X_train_text)
X_test_numeric = vectorizer.transform(X_test_text)

model_spam_email = LogisticRegression() 
model_spam_email.fit(X_train_numeric, y_train)
```

```python
# new_email_text = ["Free gift! Click to win millions of dollars now."]
new_email_text = str(input('Enter Message: '))
new_email_numeric = vectorizer.transform([new_email_text]) 
prediction_spam = model_spam_email.predict(new_email_numeric)

if (prediction_spam):
    print('This email is likely spam')
else:
    print('This email is likely not spam')
```
