```python
import numpy as np
import pandas as pd
import random
```

```python
# Define State and Actions
states = np.arange(0,5)
actions = ['GREEN', 'RED']

print('States (Traffic Levels):', states)
print('Actions (Light Control):', actions)
```

```python
Q = np.zeros((len(states), len(actions)))

alpha = 0.1    # Learning Rate
gamma = 0.9    # Discount factor
epsilon = 0.2  # Exlploration Rate
episodes = 300 # Training runs

print('Q-Table Shape:', Q.shape)
```

```python
# 0 = Empty, 1 = Light, 2 = Moderate, 3 = Heavy, 4 = Very Heavy
def get_reward(traffic, action):
    # Traffic Level 3/4 (Heavy/Very Heavy) - Goal: Keep GREEN to clear congestion
    if traffic >= 3:
        if action == 'GREEN':
            return +10  # Good: clears congestion
        else:
            return -10  # Bad: creates jams
    
    # Traffic Level 0 (Empty road) - Goal: Turn RED to save energy
    elif traffic == 0:
        if action == 'RED':
            return +5   # Saves energy
        else:
            return -5   # Wastes power
            
    # Traffic Level 1/2 (Light/Moderate traffic) - Neutral
    else: # traffic is 1 or 2
        return +1       # Neutral/Slight preference for cross-traffic
        
print('Reward example (traffic=2, FILL): ', get_reward(2,'GREEN'))
```

```python
def next_traffic(traffic, action):
    if action == 'GREEN':
        traffic += random.choice([-1,0,1])
    else:
        traffic += random.choice([-1,0,1])
    return int(np.clip(traffic,0,4))

print('Next traffic example: ', next_traffic(50,'GREEN'))
```

```python
for ep in range(episodes):
    traffic = random.choice(states)
    for _ in range(50): # steps per episode
        if random.uniform(0, 1) < epsilon:
            action = random.choice(actions)
        else:
            action = actions[np.argmax(Q[traffic])]
 
        next_state = next_traffic(traffic, action)
        reward = get_reward(next_state, action)
 
        a = actions.index(action)
        best_next = np.max(Q[next_state])
        Q[traffic, a] += alpha * (reward + gamma * best_next - Q[traffic, a])
print("Training Completed")
```

```python
try:
    traffic = int(input('\nEnter Starting Traffic Level (0=Empty to 4=Heavy)'))
    if traffic<0 or traffic>4:
        raise ValueError('Traffic level out of range')
except ValueError as ve:
    print(ve)
    traffic = 2
    print('Set to default 2')
print(f'Defaulting to Moderate traffic (traffic 2): {traffic}')

for step in range(10):
    action = actions[np.argmax(Q[traffic])]
    print(f'Step {step+1}: Traffic= {traffic} -> Action= {action}')
    traffic = next_traffic(traffic, action)
```
