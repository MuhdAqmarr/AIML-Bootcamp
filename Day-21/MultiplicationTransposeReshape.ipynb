## Matrix Multiplication
** We will use tf.matmul() method
*Rules*


1.   Inner dimensions must match
     (m,n)@(n,p)
     for e.g.


*   (5,7)@(8,9) It will not work because n are different
*   (5,8)@(8,9) It will work because inner dimension is same


2.   The result matrix has the shape of outer dimension,
     means
     A (m,n)
     B (n,p)
     So result has (m,p)
     e.g (4,7)@(7,5)->(4,5)



```python
import tensorflow as tf
import numpy as np
```

```python
tensor_A = tf.constant([[1,2],[3,4],[4,5]], dtype=tf.int32)
tensor_B = tf.constant([[5,6],[7,8]], dtype=tf.int32)
print('Shape of A', tensor_A.shape)
print('Shape of B', tensor_B.shape)
```

```python
tensor_C = tf.matmul(tensor_A, tensor_B)
print(tensor_C)
print('Shape of C', tensor_C.shape)
```

```python
X = tf.constant([
    [1.,0, -1.],
    [2.,3.,1.]
], dtype=tf.float32)
X
```

```python
Y = tf.constant([
    [4.],
    [5.],
    [6.]
], dtype=tf.float32)
Y
```

```python
print('X-Shape', X.shape)
print('Y-Shape', Y.shape)
```

```python
Z = tf.matmul(X, Y)
print(Z)
print('Shape of Z', Z.shape)
```

```python
print(f'X-Shape: {X.shape}, Y-Shape: {Y.shape}, Z-Shape: {Z.shape}')
print(X.numpy())
print(Y.numpy())
print(Z.numpy())
```

## Transpose
Transpose swaps axes. For 2d matrix shape (m,n) become (n,m) after transpose

```python
tensor_one = tf.constant([
    [[1,2],[2,3]],
    [[4,5],[6,7]]
], dtype=tf.int32)
print(tensor_one)
print('\nShape of tensor_one', tensor_one.shape)
```

```python
tensor_one_transpose = tf.transpose(tensor_one)
print(tensor_one_transpose)
print('\nShape of tensor_one_transpose', tensor_one_transpose.shape)
```

```python
# 3D Transpose with explcit axis permutation
tensor_two = tf.constant([
    [[1,2],[3,4]],
    [[5,6],[7,8]]
], dtype=tf.int32)

print(tensor_two)
```

```python
t2_transpose = tf.transpose(tensor_two, perm=[0,2,1])
print(t2_transpose)
```

## Reshape
- tf.reshape(tensor, new_shape)
- Change the shape whithout changing the data.
- *Rule*: The total number of elements must stay the same

```python
tensor_three = tf.constant([
    [1,2,3],
    [4,5,6]
], dtype=tf.int32)
print(tensor_three)
print('Shape of tensor_three', tensor_three.shape)
```

```python
tensor_three_reshape = tf.reshape(tensor_three, shape=(3,2))
print(tensor_three_reshape)
print('Shape of tensor_three_reshape', tensor_three_reshape.shape)
print('Number of elements are: ',tf.size(tensor_three_reshape).numpy())
```

```python
# Infer dimension
tensor_three_reshape_infer = tf.reshape(tensor_three, shape=(-1,2))
print(tensor_three_reshape_infer)
print('Shape of tensor_three_reshape_infer', tensor_three_reshape_infer.shape)
print('Number of elements are: ',tf.size(tensor_three_reshape_infer).numpy())
```

```python
# Infer dimension
tensor_three_reshape_infer2 = tf.reshape(tensor_three, shape=(3,-1))
print(tensor_three_reshape_infer2)
print('Shape of tensor_three_reshape_infer2', tensor_three_reshape_infer2.shape)
print('Number of elements are: ',tf.size(tensor_three_reshape_infer2).numpy())
```
