```python
import numpy as np
numpy_Array=np.arange(1,25,dtype=np.int16)
numpy_Array
```

```python
tensor_numpy=tf.constant(numpy_Array,shape=(2,3,4))
tensor_numpy
```

```python
print("Tensor Information")
print("Shape:",tensor_numpy.shape)
print("Dimension:",tensor_numpy.ndim)
print("Rank:",tf.rank(tensor_numpy).numpy())
print("Size:",tf.size(tensor_numpy))
print("Data Type of elements:",tensor_numpy.dtype)
print("Element along axis 0:",tensor_numpy.shape[0])
print("Element along axis 1:",tensor_numpy.shape[1])
print("Element along axis 2:",tensor_numpy.shape[2])
print("Total number of element: \t",tf.size(tensor_numpy).numpy())
```

```python
tensor_1=tf.constant([[11,9],[2,5]])
tensor_1
```

```python
#tensor_1+15
tf.add(tensor_1,15)
```

```python
#tensor_1-5
tf.subtract(tensor_1,15)
```

```python
#tensor_1/2
tf.divide(tensor_1,15)
```

```python
# Exercise
# Create two tensors with random values between 0 and 1 , shape is [6,300]
# **Find out the following details: **
# Number of elements
# Size,
# Rank,
# Shape

tensor_random_1 = tf.random.uniform(shape=(6, 300), minval=0, maxval=1)
tensor_random_2 = tf.random.uniform(shape=(6, 300), minval=0, maxval=1)

print("Tensor 1 Information")
print("Shape:", tensor_random_1.shape)
print("Dimension:", tensor_random_1.ndim)
print("Size:", tf.size(tensor_random_1).numpy())
print("Rank:", tf.rank(tensor_random_1).numpy())
print("Number of elements:", tf.size(tensor_random_1).numpy())

print("\nTensor 2 Information")
print("Shape:", tensor_random_2.shape)
print("Dimension:", tensor_random_2.ndim)
print("Size:", tf.size(tensor_random_2).numpy())
print("Rank:", tf.rank(tensor_random_2).numpy())
print("Number of elements:", tf.size(tensor_random_2).numpy())
```
